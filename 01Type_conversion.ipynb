Type Conversion : implicit and explicit

```python
print(5+5.6)
print(type(5) , type(5.5)) #example of implicit type conversion
```

```python
#Explicit type conversion
#string -> integer
int('4')
type(int("4"))
```

```python
int('4.5') # cannt change float to integer
```

```python
int(4+5j) #cannot change complex to integer
```

```python
#Can convert integer to string also
str(5)
type(str(5))
```

Literal ( the value given to variable is called as literal)

```python
a = 0b1010 #to write in binary use 0b
print(a)
```

```python
b = 100  #( decimal literals)
print(b)
```

```python
c = 0o310  #octal literal
print(c)
```

```python
#hexadecimal representation
d = 0x12c
print(d)
```

Float Literal

```python
float_1 = 10.5
print(float_1)
```

```python
float_2 = 1.5e2 #1.5 * 10^2
print(float_2)
```

```python
float_3 = 1.5e-3  # 1.5 * 10^-3 , minus is used to represent decimals
print(float_3)
```

```python
#complex literal
x = 3.14j
print(x)
print(x.real , x.imag)
```
